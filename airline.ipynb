```python
order = df["airline"].value_counts().sort_values(ascending=False).index    #Sort the counts,make category names as index 
```

```python
plt.figure(figsize=(7,5))
a = sns.countplot(x="airline", data=df, order=order, hue="airline", legend=False)

plt.title("Number of Flights by Airline")
plt.xlabel("Airline")
plt.ylabel("Number of Flights")

for i in a.containers:
    a.bar_label(i)

plt.show()
```

```python
order = df.groupby("airline")["price"].mean().sort_values().index   
```

```python
plt.figure(figsize=(7,5))
plt.grid()
sns.barplot(x='airline',y='price',data=df, order=order, errorbar=None )
plt.title("Price of Airline")
plt.show()
```

```python
q1=df['price'].quantile(0.65)
q3=df['price'].quantile(0.85)
```

```python
q1
```

```python
q3
```

```python
IQR=q3-q1
```

```python
IQR
```

```python
upper_limit=q3+0.5*IQR
lower_limit=q1-0.5*IQR
```

```python
upper_limit
```

```python
lower_limit
```

```python
outliers=df[(df['price']>upper_limit)|(df['price']<lower_limit)]
outliers.info()
```

```python
non_outliers_data=df[(df['price']<upper_limit)&(df['price']>lower_limit)]
non_outliers_data.info()
```

```python
sns.boxplot(x='class',y='price',data=non_outliers_data,
            hue='class')
```
